```python
def profile_dataframe(df, table_name):

    profile = pd.DataFrame({
        "Table_Name": table_name,
        "Column_Name": df.columns,
        "Data_Type": df.dtypes.astype(str).values,
        "Row_Count": len(df),
        "Missing_Count": df.isna().sum().values,
        "Missing_Percentage": (
            df.isna().sum().values / len(df) * 100
        ),
        "Unique_Count": df.nunique().values
    })

    return profile
print("Profiling DataFrames...")
for name, df in excel_data.items():
    
    if name in ["Data_Dictionary"]:
        continue

    profile = profile_dataframe(df, name)
    profile.to_excel(OUTPUT_DIR / f"{name}_profile.xlsx", index=False)
    print(f"Profile for {name} saved to {OUTPUT_DIR / f'{name}_profile.xlsx'}")
    
```

CREATE A DATA QUALITY REPORT

```python
def check_duplicates(df, key_column):

    duplicate_count = df[key_column].duplicated().sum()

    return duplicate_count
```

```python
quality_results = []

primary_keys = {
    "Customers": "Customer_ID",
    "Categories": "Category_ID",
    "Products": "Product_ID",
    "Orders": "Order_ID",
    "Order_Items": "Order_Item_ID",
    "Payments": "Payment_ID",
    "Shipping": "Shipping_ID",
    "Reviews": "Review_ID"
}
print("Checking for duplicates in primary key columns...")
for table_name, key_column in primary_keys.items():
    df = excel_data[table_name]
    duplicate_count = check_duplicates(df, key_column)
    quality_results.append({
        "Table_Name": table_name,
        "Primary_Key_Column": key_column,
        "Duplicate_Count": duplicate_count
    })
    
quality_results_df = pd.DataFrame(quality_results)
quality_results_df
```

```python
for table, key in primary_keys.items():

    df = excel_data[table]

    duplicate_count = check_duplicates(
        df,
        key
    )

    quality_results.append({
        "Table": table,
        "Primary_Key": key,
        "Duplicate_Count": duplicate_count,
        "Status": "PASS" if duplicate_count == 0 else "FAIL"
    })

quality_df = pd.DataFrame(quality_results)
print("Data Quality Check Results:")
print(quality_df)
```

CREATE BUSINESS RULE VALIDATION

```python
print("Total Products:", len(products))
# Price validation: selling price should not exceed MRP
products["Price_Check"] = np.where(
    products["Selling_Price"] <= products["MRP"],
    "PASS",
    "FAIL"
)

print("Passed:", (products["Price_Check"] == "PASS").sum())
print("Failed:", (products["Price_Check"] == "FAIL").sum())

print("\n===== FAILED RECORDS =====")

failed_price_check = products[
    products["Price_Check"] == "FAIL"
]

if len(failed_price_check) == 0:
    print("No failed records found.")
else:
    print(
        failed_price_check[
            [
                "Product_ID",
                "Product_Name",
                "Selling_Price",
                "MRP",
                "Price_Check"
            ]
        ].to_string(index=False)
    )
```

```python
# Cost vs Selling Price validation
products["Cost_Selling_Price_Check"] = np.where(
    products["Cost_Price"] < products["Selling_Price"],
    "PASS",
    "FAIL"
)

# Display results
print("===== COST VS SELLING PRICE VALIDATION =====")
print("Total Products:", len(products))
print(
    "Passed:",
    (products["Cost_Selling_Price_Check"] == "PASS").sum()
)
print(
    "Failed:",
    (products["Cost_Selling_Price_Check"] == "FAIL").sum()
)

print("\n===== FAILED RECORDS =====")

failed_cost_price = products[
    products["Cost_Selling_Price_Check"] == "FAIL"
]

if len(failed_cost_price) == 0:
    print("No failed records found.")
else:
    print(
        failed_cost_price[
            [
                "Product_ID",
                "Product_Name",
                "Cost_Price",
                "Selling_Price",
                "Cost_Selling_Price_Check"
            ]
        ].to_string(index=False)
    )
```

```python
# Rating validation
reviews["Rating_Check"] = np.where(
    reviews["Rating"].between(1, 5),
    "PASS",
    "FAIL"
)

# Display results
print("===== RATING VALIDATION =====")
print("Total Reviews:", len(reviews))

print(
    "Passed:",
    (reviews["Rating_Check"] == "PASS").sum()
)

print(
    "Failed:",
    (reviews["Rating_Check"] == "FAIL").sum()
)

print("\n===== FAILED RECORDS =====")

failed_ratings = reviews[
    reviews["Rating_Check"] == "FAIL"
]

if len(failed_ratings) == 0:
    print("No failed records found.")
else:
    print(
        failed_ratings[
            [
                "Review_ID",
                "Customer_ID",
                "Product_ID",
                "Rating",
                "Rating_Check"
            ]
        ].to_string(index=False)
    )
```

```python
# Quantity validation
order_items["Quantity_Check"] = np.where(
    (order_items["Quantity"] > 0) &
    (order_items["Quantity"] == order_items["Quantity"].astype(int)),
    "PASS",
    "FAIL"
)

# Display results
print("===== QUANTITY VALIDATION =====")
print("Total Order Items:", len(order_items))

print(
    "Passed:",
    (order_items["Quantity_Check"] == "PASS").sum()
)

print(
    "Failed:",
    (order_items["Quantity_Check"] == "FAIL").sum()
)

print("\n===== FAILED RECORDS =====")

failed_quantity = order_items[
    order_items["Quantity_Check"] == "FAIL"
]

if len(failed_quantity) == 0:
    print("No failed records found.")
else:
    print(
        failed_quantity[
            [
                "Order_Item_ID",
                "Order_ID",
                "Product_ID",
                "Quantity",
                "Quantity_Check"
            ]
        ].to_string(index=False)
    )
```

DATA QUALITY

```python
primary_keys = {
    "Customers": "Customer_ID",
    "Categories": "Category_ID",
    "Products": "Product_ID",
    "Orders": "Order_ID",
    "Order_Items": "Order_Item_ID",
    "Payments": "Payment_ID",
    "Shipping": "Shipping_ID",
    "Reviews": "Review_ID"
}

quality_results = []

for table_name, df in excel_data.items():

    if table_name == "Data_Dictionary":
        continue

    primary_key = primary_keys.get(table_name)

    duplicate_count = 0

    if primary_key in df.columns:

        duplicate_count = df[primary_key].duplicated(
            keep=False
        ).sum()

    quality_results.append({
        "Table_Name": table_name,
        "Total_Rows": len(df),
        "Total_Columns": len(df.columns),
        "Primary_Key": primary_key,
        "Duplicate_PK_Rows": duplicate_count,
        "Missing_Cells": int(df.isna().sum().sum()),
        "Status": (
            "PASS"
            if duplicate_count == 0
            else "FAIL"
        )
    })

data_quality = pd.DataFrame(quality_results)

print(data_quality)
```

FOREIGN KEY VALIDATION

```python
fk_checks = []

def validate_fk(
    child_df,
    child_table,
    child_column,
    parent_df,
    parent_table,
    parent_column
):

    valid_values = set(
        parent_df[parent_column].dropna()
    )

    invalid_rows = child_df[
        ~child_df[child_column].isin(valid_values)
        & child_df[child_column].notna()
    ]

    return {
        "Child_Table": child_table,
        "Foreign_Key": child_column,
        "Parent_Table": parent_table,
        "Parent_Key": parent_column,
        "Invalid_Rows": len(invalid_rows),
        "Status": (
            "PASS"
            if len(invalid_rows) == 0
            else "FAIL"
        )
    }
```

```python
fk_checks.append(
    validate_fk(
        products,
        "Products",
        "Category_ID",
        categories,
        "Categories",
        "Category_ID"
    )
)

fk_checks.append(
    validate_fk(
        orders,
        "Orders",
        "Customer_ID",
        customers,
        "Customers",
        "Customer_ID"
    )
)

fk_checks.append(
    validate_fk(
        order_items,
        "Order_Items",
        "Order_ID",
        orders,
        "Orders",
        "Order_ID"
    )
)

fk_checks.append(
    validate_fk(
        order_items,
        "Order_Items",
        "Product_ID",
        products,
        "Products",
        "Product_ID"
    )
)

fk_checks.append(
    validate_fk(
        payments,
        "Payments",
        "Order_ID",
        orders,
        "Orders",
        "Order_ID"
    )
)

fk_checks.append(
    validate_fk(
        shipping,
        "Shipping",
        "Order_ID",
        orders,
        "Orders",
        "Order_ID"
    )
)

fk_checks.append(
    validate_fk(
        reviews,
        "Reviews",
        "Customer_ID",
        customers,
        "Customers",
        "Customer_ID"
    )
)

fk_checks.append(
    validate_fk(
        reviews,
        "Reviews",
        "Product_ID",
        products,
        "Products",
        "Product_ID"
    )
)

fk_checks.append(
    validate_fk(
        reviews,
        "Reviews",
        "Order_ID",
        orders,
        "Orders",
        "Order_ID"
    )
)

fk_validation = pd.DataFrame(fk_checks)

print(fk_validation)
```

KPI Summary

```python
total_revenue = orders["Order_Total"].sum()

total_orders = orders["Order_ID"].nunique()

total_customers = customers["Customer_ID"].nunique()

total_products = products["Product_ID"].nunique()

total_quantity = order_items["Quantity"].sum()

total_profit = order_items["Profit_Amount"].sum()

average_order_value = (
    total_revenue / total_orders
    if total_orders > 0
    else 0
)

cancelled_orders = (
    orders["Order_Status"]
    .eq("Cancelled")
    .sum()
)

cancellation_rate = (
    cancelled_orders / total_orders
    if total_orders > 0
    else 0
)

returned_orders = (
    orders["Order_Status"]
    .eq("Returned")
    .sum()
)

return_rate = (
    returned_orders / total_orders
    if total_orders > 0
    else 0
)

average_rating = reviews["Rating"].mean()

average_delivery_days = shipping[
    "Delivery_Days"
].mean()

successful_payments = (
    payments["Payment_Status"]
    .eq("Successful")
    .sum()
)

payment_success_rate = (
    successful_payments / len(payments)
    if len(payments) > 0
    else 0
)

profit_margin = (
    total_profit / total_revenue
    if total_revenue != 0
    else 0
)
```

```python
kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Revenue",
        "Total Profit",
        "Profit Margin %",
        "Total Orders",
        "Total Customers",
        "Total Products",
        "Total Quantity Sold",
        "Average Order Value",
        "Cancellation Rate %",
        "Return Rate %",
        "Average Rating",
        "Average Delivery Days",
        "Payment Success Rate %"
    ],

    "Value": [
        total_revenue,
        total_profit,
        profit_margin * 100,
        total_orders,
        total_customers,
        total_products,
        total_quantity,
        average_order_value,
        cancellation_rate * 100,
        return_rate * 100,
        average_rating,
        average_delivery_days,
        payment_success_rate * 100
    ]
})

print(kpi_summary)
```

Category Analysis

```python
category_data = (
    order_items
    .merge(
        products[
            [
                "Product_ID",
                "Category_ID"
            ]
        ],
        on="Product_ID",
        how="left"
    )
    .merge(
        categories[
            [
                "Category_ID",
                "Category_Name"
            ]
        ],
        on="Category_ID",
        how="left"
    )
)
```

```python
category_analysis = (
    category_data
    .groupby(
        [
            "Category_ID",
            "Category_Name"
        ],
        as_index=False
    )
    .agg(
        Quantity_Sold=("Quantity", "sum"),
        Revenue=("Line_Total", "sum"),
        Profit=("Profit_Amount", "sum"),
        Order_Items=("Order_Item_ID", "count")
    )
)
```

```python
category_analysis["Profit_Margin_%"] = np.where(
    category_analysis["Revenue"] != 0,
    (
        category_analysis["Profit"]
        /
        category_analysis["Revenue"]
    ) * 100,
    0
)
```
